## Import Dependencies

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
```

```python
# Loading the dataset : 
df = pd.read_csv('Cars_price.csv')
df
```

```python
df.columns
```

```python
# Preprocessing : 
df.columns = df.columns.str.strip()
print(df.columns)
```

```python
# Lets add a row : 
df.loc[5] = [3,50000,None,2100000]
```

```python
df
```

```python
df.describe()
```

```python
mean_year = int(df['year'].mean())
df['year'] = df['year'].fillna(float(mean_year))
```

```python
df
```

```python
# Training the model : 
reg = linear_model.LinearRegression()
reg.fit(df[['cylinders', 'milage', 'year']], df.price)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[4,30000,2021]])
```
