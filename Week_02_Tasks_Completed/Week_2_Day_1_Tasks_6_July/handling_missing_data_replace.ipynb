## <font color="maroon"><h4 align="center">Handling Missing Data - replace method</font>

```python
import pandas as pd
import numpy as np

df = pd.read_csv('weather_data.csv')
df
```

## Replacing Single Value 

```python
new_df = df.replace(-99999, value=np.nan)
new_df
```

## Replacing List with a Single Value

```python
new_df = df.replace(to_replace=[-99999, -88888], value=0)
new_df
```

## Replacing Per Column

```python
new_df = df.replace({
    'temperature' : -99999,
    'windspeed' : -99999, 
    'event' : '0'
}, value=np.nan)

new_df 
```

## Replacing By Using Mapping

```python
new_df = df.replace({
    -99999 : np.nan,
    'no event' : 'Sunny',
})

new_df
```

```python
new_df
```

```python
new_df = df.replace({
    'event' : '0'
}, value=np.nan)

new_df
```

```python
new_df = new_df.replace({
    'temperature' : -99999,
    'windspeed' : -99999,
}, value=0)

new_df
```

```python
new_df = new_df.replace({
    'event' : np.nan
},value='Sunny')

new_df 
```

## Regex

```python
# When wind speed is 6mph, 7mph etc and temperature is 32 F and 28 F etc.
new_df = df.replace({'temperature' : '[A-Za-z]', 'windspeed':'[a-z]'}, '', regex=True)
new_df
```
