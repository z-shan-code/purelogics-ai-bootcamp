# <font color="blue"><h4 align="left">Basic Concatenation</h3></font>

```python
import pandas as pd
```

```python
pakistani_weather = pd.DataFrame({
    'city':['karachi', 'lahore', 'peshawar'],
    'temperature' : [32,45,30],
    'humidity' : [80, 60,78]
})
```

```python
pakistani_weather
```

```python
us_weather = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68, 65, 75]
})
us_weather
```

```python
df = pd.concat([pakistani_weather, us_weather])
```

```python
df
```

## Ignore Index

```python
df = pd.concat([pakistani_weather, us_weather], ignore_index=True)
df
```

```python
df = pd.concat([pakistani_weather,us_weather], keys=['Pakistan', 'US'])
df
```

```python
df.loc['US']
```

```python
df.loc['Pakistan']
```

## Concatenation Using Index

```python
temperature_df = pd.DataFrame({
    'city' : ['karachi', 'lahore', 'peshawar'],
    'temperature' : [32,45,30]
}, index=[1,2,3])

temperature_df
```

```python
windspeed_df = pd.DataFrame({
    "city": ["lahore","karachi"],
    "windspeed": [7,12],
}, index=[1,0])
windspeed_df
```

```python
new_df = pd.concat([temperature_df, windspeed_df],axis=1)
```

```python
new_df
```

## Concatenate DataFrame with Series

```python
s = pd.Series(['Humid', 'Dry', 'Rain'], name='event')
s
```

```python
pd.concat([temperature_df, s],axis=1)
```
