# <font color="purple"><h3 align="center">Pandas Concatenate Tutorial</h3></font>

## <font color='blue'>Basic Concatenation</font>

```python
import pandas as pd

pakistani_weather = pd.DataFrame({
    "city": ["karachi","lahore","peshawar"],
    "temperature": [32,45,30],
    "humidity": [80, 60, 78]
})
pakistani_weather
```

```python
us_weather = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68, 65, 75]
})
us_weather
```

```python
# concat us and pakistani_weather

df = pd.concat([pakistani_weather, us_weather], ignore_index=True)
df
```

## <font color='blue'>Concatenation And Keys</font>

```python
df = pd.concat([pakistani_weather, us_weather], keys=["pakistan", "us"])
df
```

```python
# Show cities temperature, humidity of US cities
df.loc['us']
```

```python
# Show cities temperature, humidity of Pakistani cities
df.loc['pakistan']
```

## <font color='blue'>Concatenation Using Index</font>

```python
temperature_df = pd.DataFrame({
    "city": ["karachi","lahore","peshawar"],
    "temperature": [32,45,30],
}, index=[0,1,2])
temperature_df
```

```python
windspeed_df = pd.DataFrame({
    "city": ["lahore","karachi"],
    "windspeed": [7,12],
}, index=[1,0])
windspeed_df
```

```python
#concatenate the temperature and windspeed of Lahore, Karachi and Peshawar

pd.merge(temperature_df, windspeed_df, how='left')
```

## <font color='blue'>Concatenate dataframe with series</font>

```python
s = pd.Series(['Humid', 'Dry', 'Rain'], name='event')
s
```

```python
df = pd.concat([temperature_df, s], axis=1)
```

```python
df
```

### For this dataset, get following answers,
#### 1. What was the maximum temperature in each of these 3 cities?
#### 2. What was the average windspeed in each of these 3 cities?


```python
df.groupby('city')['temperature'].max()
```

```python
df.groupby('city')['windspeed'].mean()
```

```python
g = df.groupby('city')
g
```

```python
for city, data in g : 
    print('\ncity', city)
    # print('\n')
    print('data', data)
```

# <font color="purple"><h3 align="center">Pandas Merge Tutorial</h3></font>

```python
import pandas as pd
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","orlando"],
    "humidity": [65,68,75],
})
df2
```

```python
pd.merge(df1,df2)
```

## <font color='blue'>Type Of DataBase Joins</font>

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35, 38],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "humidity": [65,68,71],
})
df2
```

```python
df3 = pd.merge(df1,df2,on='city', how='inner')
df3
```

```python
df3 = pd.merge(df1,df2,on='city', how='outer')
df3
```

```python
df3 = pd.merge(df1,df2,on='city', how='left')
df3
```

```python
df3 = pd.merge(df1,df2,on='city', how='right')
df3
```

## <font color='blue'>indicator flag</font>

```python
df3 = pd.merge(df1,df2,on='city', how='outer',indicator=True)
df3
```

## <font color='blue'>suffixes</font>

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35,38],
    "humidity": [65,68,71, 75]
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "temperature": [21,14,35],
    "humidity": [65,68,71]
})
df2
```

```python
df3= pd.merge(df1,df2,on="city",how="outer", suffixes=('_first','_second'))
df3
```
