## a) : Reshape into matrix form

Assume each row is a day

Reshape temperature into a (10,24) matrix

Calculate daily min, max, mean across rows

```python
# Extract temperature and wind as Numpy arrays
import numpy as np

temp = df['temperature_C'].to_numpy()
wind = df['wind_speed_kmph'].to_numpy()
```

```python
# Reshaping temperature into a 10,24 matrix

temp_matrix = temp.reshape((10,24))

# function for finding min : 
def min() :
    return temp_matrix.min()

# function for finding max : 
def max() :
    return temp_matrix.max()

# function for finding min : 
def mean() :
    return temp_matrix.mean()


print(f'Minimum value for temp col : {min()}')    
print(f'Maximum value for temp col : {max()}')    
print(f'Mean value for temp col : {mean()}') 
```

### b) Normalize the temperature matrix
- Subtract the mean and divide by std deviation
- Do it manually using NumPy functions

```python
# Finding the mean of the whole temp_matrix

mean_temp_matrix = np.mean(temp_matrix)
mean_temp_matrix
```

```python
# Finding the std deviation of temp_matrix

std_temp_matrix = np.std(temp_matrix)
std_temp_matrix
```

```python
# Take every value one-by-one from the matrix and then subtract it from mean and then divide by std deviation. 

normalized_matrix = (temp_matrix - mean_temp_matrix) / std_temp_matrix
print(normalized_matrix)
```

### c) Apply custom mask/filter
- Create a mask for wind speed > 15 kmph
- Use it to extract high-wind readings

```python
# Step 1: Create the mask
wind_mask = wind > 15
print(wind_mask)
```

```python
# Step 2: Apply the mask to extract high wind readings
high_wind_readings = wind[wind_mask]
print(high_wind_readings)
```

## Final Challenge: Write Your Own Function
Write a function `daily_summary(matrix)` that takes a NumPy matrix of shape (10, 24) and returns a summary dictionary for each day.

```python
def daily_summary(matrix) :
    summary_dict = []
    
    for i in matrix : 
        day_dict = {
            'mean' : i.mean(), 
            'max' : i.max(), 
            'min' : i.min() 
        }

        summary_dict.append(day_dict)
    return summary_dict
    
summaries = daily_summary(temp_matrix)
```

```python
for day, summary in enumerate(summaries, 1) : 
    print(f'Day : {day}, Mean : {summary['mean']:.2f}, Max : {summary['max']:.2f}, Min : {summary['min']:.2f}')
```
