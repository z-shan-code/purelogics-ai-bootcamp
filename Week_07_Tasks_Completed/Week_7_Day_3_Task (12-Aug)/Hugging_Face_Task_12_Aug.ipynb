## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

### 2.3 Question Answering

```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
context = (
    "Hugging Face is an AI company with the mission to democratize good machine learning. "
    "The Transformers library provides thousands of pretrained models in 100+ languages."
)

answer = qa({
    "question": "What is the mission of Hugging Face?",
    "context": context
})
print(answer["answer"])
```

# Import pipeline

````python
from transformers import pipeline

# pipeline is a high-level wrapper around the model
# Instead of manually doing the :
# text -> tokenizer -> model-> logits -> probabilities -> label
# Hugging Face describes a pipeline as combining preprocessing, model inference, and post-processing
````

# Creating our Sentiment Pipeline

````python
sentiment_pipeline = pipeline(
    'sentiment-analysis',
    model='distilbert/distilbert-base-uncased-finetuned-sst-2-english'
)

# There are 2 main things :
# First we tell what task to perform and then in the model parameter, we mention the pre-trained model that we want to use.
# The model has already been fine-tuned on the SST-2 sentiment classification task and produces POSITIVE or NEGATIVE predictions
````

# Testing on 3 Product reviews :

````python
reviews = [
    "This phone is amazing. The camera quality is excellent and the battery lasts all day.",
    "The headphones stopped working after two days. Very disappointing purchase.",
    "The product is okay, but I expected better quality for the price."
]

for review in reviews :
  result = sentiment_pipeline(review)
  print("Review : ", review)
  print("Result : ", result)
  print("-" * 80)


print(f"\n\nThe label '{result[0]['label']}' tells you the direction of sentiment; the score '{result[0]['score']}' tells you how confident the model is about that direction.")
````

# Experiment 1 : Max Length

````python
long_review = """
I bought this laptop expecting excellent performance because the specifications
looked very impressive. However, after using it for several weeks, I noticed
that the battery drains quickly, the laptop becomes very hot, and the overall
performance is not as smooth as I expected.
"""

result_1 = sentiment_pipeline(long_review, truncation=True, max_length=6)
result_1
````

````python
result_2 = sentiment_pipeline(long_review, truncation=True, max_length=50)
result_2
````

# Another Comparison

````python
review = """
The phone looks beautiful and has an excellent camera, but the battery
drains extremely quickly and the device becomes hot during normal use.
"""

result_15 = sentiment_pipeline(
    review,
    truncation=True,
    max_length=15
)

result_200 = sentiment_pipeline(
    review,
    truncation=True,
    max_length=200
)

print("MAX LENGTH = 15")
print(result_15)

print("\nMAX LENGTH = 200")
print(result_200)
````

### Parameter Experimentation

For this experiment, I used a sentiment-classification pipeline with a pretrained DistilBERT model. The task instructions mentioned parameters such as `max_length`, `top_k`, and `temperature`. However, `top_k` and `temperature` are primarily used for **text generation**, where a model generates new tokens and these parameters control the selection and randomness of generated text.

Since sentiment classification does not generate text but instead predicts a sentiment label (`POSITIVE` or `NEGATIVE`), these parameters are not directly applicable to this pipeline. Therefore, I experimented with `max_length`, which is relevant to classification because it controls the maximum number of input tokens processed by the model.

The results showed that changing `max_length` can affect how much of a long input is considered by the model and can potentially influence its prediction and confidence.

### Conclusion

The pretrained DistilBERT model successfully classified product reviews as positive or negative with high confidence. The experiment also demonstrated that the `score` represents the model's confidence in its predicted label, rather than how positive or negative the review is. Overall, the Hugging Face `pipeline()` provided a simple way to apply a pretrained NLP model without manually implementing tokenization, model inference, and post-processing.
